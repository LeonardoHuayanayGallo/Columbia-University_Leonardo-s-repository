```python
#Exercise 2
def getBondPrice(y, face, couponRate, m, ppy=1):
    c = face * couponRate / ppy #coupon per period
    periods = m * ppy #total number of periods
    bondPrice = 0.0 #accumulator for PV of all CFs

    #Loop through each coupon payment
    for t in range(1, periods + 1):
        bondPrice += c / (1 + y/ppy)**t #discounted coupon

    #Add discounted principal at maturity
    bondPrice += face / (1+ y/ppy)**periods

    return bondPrice

# Test values
y = 0.03
face = 2000000
couponRate = 0.04
m = 10

#Case 1: Annual payments (ppy=1)
print("Annual payments:", getBondPrice(y, face, couponRate, m, 1))

#Case 2: semiannual payments (ppy=2)
print("Semiannual payments:", getBondPrice(y, face, couponRate, m, 2))

#Case 3: default ppy (should assume 1)
print("Default payments:", getBondPrice(y, face, couponRate, m))
        
```

```python
#Exercise 3
def getBondDuration(y, face, couponRate, m, ppy=1):
    c = face * couponRate / ppy #coupon payment per period
    periods = m * ppy #total number of periods

    #compute bond price using exercise 2 function
    bondPrice = getBondPrice(y, face, couponRate, m, ppy)

    #weighted sum of time * PV(cash flow)
    weighted_sum = 0.0
    for t in range(1, periods + 1):
        #cash flow: coupon each period, plus face value at maturity
        cf = c if t < periods else c + face
        #Present value of this cash flow
        pv = cf / (1 + y/ppy)**t
        #Add weighted contribution
        weighted_sum += t * pv

    #divide numerator by price to get Macaulay duration
    bondDuration = weighted_sum / bondPrice

    return bondDuration

#Test values
y = 0.03
face = 2000000
couponRate = 0.04
m = 10
ppy = 1

print("Duration (annual):", round(getBondDuration(y, face, couponRate, m, ppy), 2))
```

```python
#Exercise 4
def getBondPrice_enumerate(face, couponRate, m, yc):
    c = face * couponRate #annual coupon payment
    # Build the list of cash flows: m-1 coupons and the last coupon + face
    cashflows = [c] * (m - 1) + [c + face]

    bondPrice = 0.0
    #Enumerate gives us (t, cf) where t = 1, 2, ..., m
    for t, cf in enumerate(cashflows, 1):
        bondPrice += cf / (1 + yc[t-1])**t #discount each cash flow
    return bondPrice

#Test values
yc = [0.01, 0.015, 0.02, 0.025, 0.03] #yields for years 1 to 5
face = 2000000
couponRate = 0.04
m = 5

print("Bond price (enumerate):", round(getBondPrice_enumerate(face, couponRate, m, yc), 2))
```

```python
#Exercise 5

def getBondPrice_zip(face, couponRate, times, yc):
    c = face * couponRate
    bondPrice = 0.0
    last_t = times[-1] #maturity time

    #zip combines times and yields into pairs (t, y)
    for t, y in zip(times, yc):
        cf = c if t < last_t else c + face #last time adds princpal
        bondPrice += cf / (1 + y)**t #discount with irregular exponent
    return bondPrice
#Test values
face = 2000000
couponRate = 0.04
times = [1, 1.5, 3, 4, 7]
yc = [0.01, 0.015, 0.02, 0.025, 0.03]

print("Bond price (zip):", round(getBondPrice_zip(face, couponRate, times, yc), 2))
```

```python
#Exercise 6

def FizzBuzz(start, finish):
    outlist = [] #empty list to store results
    for i in range(start, finish + 1): #loop from start to finish (inclusive)
        if i % 3 == 0 and i % 5 == 0: #divisible by both 3 and 5
            outlist.append("fizzbuzz")
        elif i % 3 == 0: #divisible by 3 only
            outlist.append("fizz")
        elif i % 5 == 0: #divisible by 5 only
            outlist.append("buzz")
        else:    # not divisible by 3 or 5
            outlist.append(str(i))
    return outlist

#Test values
print(FizzBuzz(1,15))
```
