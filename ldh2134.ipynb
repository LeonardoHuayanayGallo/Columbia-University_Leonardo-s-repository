```python
#Exercise 2
import numpy as np

def getBondPrice(y, face, couponRate, m, ppy=1):
    #Number of total payments
    n = m * ppy
    #Coupon payment amount
    c = face * couponRate / ppy
    #Periods (array from 1 to n)
    t = np.arange(1, n+1)
    #Discount factors for each period
    df = 1 / (1 + y / ppy) ** t
    #Cash  flows: coupon for all periods, face+coupon at maturity
    cf = np.full(n, c)
    cf[-1] += face

    #Present value of each cash flow
    pvcf = cf * df
    #Bond price: sum of PVs
    return pvcf.sum()

#Test values

y = 0.03
face = 2000000
couponRate = 0.04
m = 10
ppy = 1

price = getBondPrice(y, face, couponRate, m, ppy)
print(price)
```

```python
#Exercise 3
import numpy as np

def getBondDuration(y, face, couponRate, m, ppy=1):
    n = m * ppy
    c = face * couponRate / ppy
    t = np.arange(1, n+1)
    df = 1 / (1 + y / ppy) ** t     #discount factors
    cf = np.full(n, c)
    cf[-1] += face     #add face value at maturity
    pvcf = cf * df   #present value of each flow

    duration = np.sum(pvcf * t) / np.sum(pvcf) #weighted average, Macaulay duration
    return duration

# Test values
y = 0.03
face = 2000000
couponRate = 0.04
m = 10
ppy = 1

print (getBondDuration(y, face, couponRate, m, ppy))
    
```

```python
#Exercise 4

import numpy as np

def FizzBuzz(start, finish):
    numvec = np.arange(start,finish + 1)
    objvec = numvec.astype(object)

    #Apply FizzBuzz logic using masks (vectorized, no loops)
    fizzbuzz_mask = (numvec % 3 == 0) & (numvec % 5 == 0)
    fizz_mask = (numvec % 3 == 0) & ~fizzbuzz_mask
    buzz_mask = (numvec % 5 == 0) & ~fizzbuzz_mask

    objvec[fizzbuzz_mask] = "fizzbuzz"
    objvec[fizz_mask] = "fizz"
    objvec[buzz_mask] = "buzz"
    return objvec
    
print(FizzBuzz(16, 30))
```
